# GFAD Aug Lite

Thin Kaggle runner for class-specific lighter augmentation on screw and capsule.

```python
from pathlib import Path
import os
import subprocess
import sys

REPO_URL = "https://github.com/luythoangduy/gfad.git"
REPO_BRANCH = "exp/screw-capsule-aug-lite"
REPO_DIR = Path("/kaggle/working/gfad")

if not REPO_DIR.exists():
    subprocess.run(["git", "clone", "--depth", "1", "--branch", REPO_BRANCH, REPO_URL, str(REPO_DIR)], check=True)

os.chdir(REPO_DIR)
print(f"Repo: {REPO_DIR}")
```

```python
INSTALL_DEPS = True

if INSTALL_DEPS:
    subprocess.run([sys.executable, "-m", "pip", "install", "hydra-core", "torchmetrics"], check=True)
```

```python
subprocess.run(
    [sys.executable, "scripts/run_mvtec_kaggle_experiment.py", "--config", "configs/mvtec_aug_lite_screw_capsule_all_classes.yaml"],
    check=True,
)
```
